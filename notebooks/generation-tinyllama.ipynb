# TinyLlama chat model trial

[Back to the introduction](demo.ipynb). Optional heavyweight experiment: use a fresh kernel. The q4 weights are about 910 MB, with additional session memory. This notebook selects WebGPU/q4, which requires compatible hardware, browser, and runtime; it does not promise compatibility or silently fall back to WASM. Change the configuration if your environment supports another combination.

`LLM.chat()` applies the chat template included in the loaded tokenizer and returns only the assistant response.

```python
from lmlite import LLM
```

```python
gen_tinyllama = await LLM.create(
    generator_model="Xenova/TinyLlama-1.1B-Chat-v1.0",
    backend="webgpu",
    dtype="q4",
    max_new_tokens=50,
    return_full_text=False,  # Return the answer without the chat prompt.
    do_sample=False,
)
```

```python
await gen_tinyllama.chat([
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is JupyterLite?"},
])
```

```python
# Print the answer as it is generated. Leaving the loop early (break, an error or a kernel
# interrupt) cancels the generation; another task can also stop it with `await gen_tinyllama.cancel()`.
chunks = []
async for chunk in gen_tinyllama.chat_stream([
    {"role": "user", "content": "Explain embeddings simply."},
]):
    print(chunk, end="", flush=True)
    chunks.append(chunk)
    if len(chunks) == 20:
        break
```

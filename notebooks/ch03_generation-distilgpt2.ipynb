# DistilGPT-2 model trial

[Back to the introduction](ch01_demo.ipynb). Run in order in a fresh kernel. This loads only DistilGPT-2, a smaller GPT-2 completion model, not an instruction-following assistant. WASM is the default; WebGPU is optional when supported. Downloaded weights are cached, but inference still needs browser memory.

```python
from lmlite import LLM
```

```python
gen_distilgpt2 = await LLM.create(
    generator_model="Xenova/distilgpt2",
    backend="wasm",
    max_new_tokens=30,
    do_sample=False,
)
```

```python
await gen_distilgpt2.generate("Bindings are useful because")
```

```python
await gen_distilgpt2.dispose()
```

The model is disposed without clearing cached downloads. Rerun the trial to load it again. Restart the kernel before a different heavyweight model trial.

# Text classification

[Back to the introduction](ch01_demo.ipynb). Run in order. This notebook loads one English sentiment classifier, independently of generation and embeddings. Its labels are POSITIVE/NEGATIVE, with no neutral class; results are model predictions, not general-purpose judgments. `backend` and `dtype` configure classification, while embedding-specific options do not.

```python
from lmlite import LLM
```

```python
llm_cls = await LLM.create(
    classification_model="distilbert-base-uncased-finetuned-sst-2-english",
    backend="wasm",
)
```

```python
texts = ["I love using lmlite in JupyterLite!", 
         "This workflow is frustrating and slow."]
for text in texts:
    print(text)
    for label in await llm_cls.classify(text, top_k=2):
        print(f"  {label['label']}: {label['score']:.4f}")
```

```python
await llm_cls.dispose()
```

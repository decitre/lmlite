# LaMini-Neo model trial

[Back to the introduction](ch01_demo.ipynb). Optional model compatibility experiment, isolated from the main demo. Use a fresh kernel. Availability depends on the model's ONNX exports and Transformers.js runtime. This preserves the original WebGPU trial; unsupported configurations fail explicitly. Generated answers are not guaranteed factual.

```python
from lmlite import LLM
```

```python
gen_lamini = await LLM.create(
    generator_model="Xenova/LaMini-Neo-125M",
    backend="webgpu",
    max_new_tokens=40,
    return_full_text=False,  # Return the answer without the instruction prompt.
    do_sample=False,
)
```

```python
await gen_lamini.generate("Explain why Python bindings are useful.")
```

```python
await gen_lamini.dispose()
```

# Speak to the notebook

[Back to the introduction](ch01_demo.ipynb). Run in order. This notebook records your voice with the device's microphone and transcribes it in the browser, so nothing leaves the page. It works on a laptop or a tablet: the page must be served over HTTPS (or localhost), and the browser asks for microphone permission the first time you tap Record.

A notebook kernel runs in a Web Worker, which cannot open the microphone. The `AudioRecorder` widget therefore records on the page itself and hands the kernel a 16 kHz mono WAV file, which a speech recognition pipeline takes as bytes. Outside JupyterLite, install it with `pip install 'lmlite[audio]'`.

```python
from lmlite.audio import AudioRecorder

recorder = AudioRecorder()
recorder
```

Tap **Record**, say a sentence in English, then tap **Stop**. The next cell waits for the recording.

If the widget says the recording is silent, or you hear nothing, look first at the system settings: the microphone's **input level** (a level at 0 records pure silence, whatever the browser does), that the right input device is selected, and that the browser may use the microphone. If those are fine, tick **Raw input** and record again. By default the browser processes the microphone (echo cancellation, noise suppression, gain control), and on some devices that processing can mix the input down to silence; Raw input turns it off and levels the recording instead.

```python
from IPython.display import Audio

wav = await recorder.wait()
info = recorder.info
print(f"{info['seconds']:.1f} s of audio, {len(wav)} bytes, peak {info['peak']:.2f} "
      f"({info['channels']} channel(s), {info['sampleRate']} Hz, raw input: {info['raw']})")
if info["peak"] < 0.005:
    print("This recording is silent: check the input level and the selected input in the system settings (a level at 0 is the usual cause),")
    print("then tick Raw input if they are fine, run recorder.clear(), record again and rerun this cell.")
Audio(wav)
```

## Transcribe it

`whisper-tiny.en` is an English-only model of about 40 MB. Recordings longer than 30 seconds need `chunk_length_s=30`. Tap Record again and rerun the cells from `wav = await recorder.wait()` after `recorder.clear()` to try another sentence.

```python
from lmlite import LLM

transcriber = await LLM.pipeline("automatic-speech-recognition", model="whisper-tiny.en", backend="wasm")
result = await transcriber(wav)
print(result["text"])
await transcriber.dispose()
```

## What instrument is playing?

The same recorder can listen to music. `onnx-community/Musical-Instrument-Classification-ONNX` (about 95 MB, MIT licence) names one of nine sounds: acoustic guitar, bass guitar, drum set, electric guitar, flute, hi-hats, keyboard, trumpet and violin. Play an instrument near the microphone, or play a recording from another device, then tap **Stop** after a few seconds.

Tick **Raw input** before you record: the browser's noise suppression and gain control are made for speech and may alter music. The model is small and unsure, so its best score is often around 0.3: read the top three, and expect mistakes.

```python
music = AudioRecorder()
music
```

```python
wav = await music.wait()
classifier = await LLM.pipeline("audio-classification", model="onnx-community/Musical-Instrument-Classification-ONNX", backend="wasm")
for item in await classifier(wav, top_k=3):
    print(f"{item['label']:16s} {item['score']:.2f} {'#' * round(item['score'] * 40)}")
await classifier.dispose()
```

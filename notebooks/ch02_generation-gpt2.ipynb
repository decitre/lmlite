# GPT-2 model trial

[Back to the introduction](ch01_demo.ipynb). This notebook loads only GPT-2. Use a fresh kernel and run in order. fp32 weights are about 501 MB; browser sessions need additional memory. The default here is WASM; set `backend="webgpu"` when supported. GPT-2 continues text and is not instruction-tuned. Model/backend compatibility is runtime-dependent.

```python
from lmlite import LLM
```

```python
gen_gpt2 = await LLM.create(
    generator_model="Xenova/gpt2",
    backend="wasm",
    dtype="fp32",
    max_new_tokens=30,
    do_sample=False,
)
```

```python
await gen_gpt2.generate("Bindings are useful because")
```

```python
if False:
    # Opt in to portable ZIP export; it can require substantial memory.
    await gen_gpt2.export_models()
```

```python
await gen_gpt2.dispose()
```
